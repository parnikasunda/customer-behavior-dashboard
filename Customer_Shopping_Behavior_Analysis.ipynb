```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median())) #filling missing values with median of each seperate categories not the overall median
```

```python
df.isnull().sum()
```

```python
#snake casing for consistency
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# creating a column age_group by dividing the 'age' column into 4 equal-sized groups (quartiles) using pd.qcut()
labels = ['Young Adult', 'Adult' , 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create a column purchase_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly' : 90,
    'Bi - Weekly' : 14,
    'Annually' : 365,
    'Every 3 months': 90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
